```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import missingno as msno
```

```python
df=pd.read_csv('/Stores.csv')
```

```python
df.head()
```

```python
df.drop(columns='Store ID ', axis=1,inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
msno.bar(df,figsize=(10,4),color='red')
plt.show()
```

```python
for column_name in df.columns:
    unique_values = len(df[column_name].unique())
    print("Feature '{column_name}' has '{unique_values}' unique values".format(column_name = column_name,
                                                                                         unique_values=unique_values))
```

```python
for column_name in df.columns:
    plt.figure(figsize=(10,4))
    plt.title("distribution of {}".format(column_name))
    sns.histplot(df[column_name],stat='density',color='green')
    sns.kdeplot(df[column_name],color='black')
    plt.axvline(df[column_name].mean(),color='red',linewidth=0.8,linestyle='--')
    plt.show()
```
